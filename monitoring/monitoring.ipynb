```python
from pathlib import Path
import pandas as pd


# Si ton notebook est dans MLOps2/monitoring/,
# parent = racine MLOps2.
PROJECT_ROOT = Path("..").resolve()

DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "sample_10000_stratified.csv"
)

MONITORING_DIR = (
    PROJECT_ROOT
    / "monitoring_data"
)

MONITORING_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

OUTPUT_PATH = (
    MONITORING_DIR
    / "production_local_test.parquet"
)


print("Racine du projet :", PROJECT_ROOT)
print("CSV source :", DATA_PATH)
print("Fichier de sortie :", OUTPUT_PATH)


df_current = pd.read_csv(DATA_PATH)

# TARGET ne doit pas faire partie des features surveillées.
df_current = df_current.drop(
    columns=["TARGET"],
    errors="ignore",
)

df_current.to_parquet(
    OUTPUT_PATH,
    engine="pyarrow",
    index=False,
)

print(f"\nParquet créé avec succès : {OUTPUT_PATH}")
print(f"Dimensions : {df_current.shape}")
```
